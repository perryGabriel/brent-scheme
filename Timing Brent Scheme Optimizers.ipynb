```python
iterations = 10
scheme = BrentScheme(n=4,d=4,m=4,p=48)

print(f"Timing {iterations} iterations of each function:\n")

# stepper.optimize(scheme, batch_size=1)
# slowest optimizer
start_time = time.time()
for _ in trange(iterations):
    try:
        stepper.optimize(scheme, batch_size=1)
    except:
        pass
end_time = time.time()
print(f"stepper.optimize: {end_time - start_time:.4f} seconds\n")

# stepper.epoch_pseudoinverse(scheme, batch_size=1)
start_time = time.time()
for _ in trange(iterations):
    stepper.epoch_pseudoinverse(scheme, batch_size=1)
end_time = time.time()
print(f"stepper.epoch_pseudoinverse: {end_time - start_time:.4f} seconds\n")

# stepper.epoch(scheme, penalty=0.2, lr=5e-5, batch_size=1)
# fasted optimizer
start_time = time.time()
for _ in trange(iterations):
    stepper.epoch(scheme, penalty=0.2, lr=5e-5, batch_size=1)
end_time = time.time()
print(f"stepper.epoch: {end_time - start_time:.4f} seconds\n")

# manipulator.normalize(scheme)
start_time = time.time()
for _ in trange(iterations):
    manipulator.normalize(scheme)
end_time = time.time()
print(f"\nmanipulator.normalize: {end_time - start_time:.4f} seconds")
```
